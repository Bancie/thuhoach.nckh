## Bottleneck point

### Description

For $v \in V, \: x = (e, \alpha) \in G$, the bottleneck point $v$ on $e=(v_i,v_j)$ is

$$
\alpha l_e + d(v_i, v) = (1-\alpha ) l_e + d(v_j, v)
$$

- $BP_e$: Set of all bottleneck points on $e \in E$.

- $BP$: Set of all bottleneck points on $G$.

### Example

```python
graph = [
    [0, 1, 3],
    [1, 0, 2],
    [3, 2, 0]
]
```

```python
ShowGraph = Graph(graph)
ShowGraph.show()
```

```python
from packages.Point import Local
```

```python
a = Local(graph, 3, 0, 2)
print(a.BottleNeck(1))
```

## Equilibrium point

### Description

For $v_k, v_l \in V, \: v_k \neq v_l, \: x = (e, \alpha) \in G, \: e= (v_u, v_v)$, the equililbrium point is

$$
\omega _k (\alpha l_e+ d(v_k, v_u)) = \omega _l ((1-\alpha )l_e + d(v_l, v_v))
$$

If $x \in G$

$$
\omega _k d(v_k, x) = \omega _l d(v_l, x)
$$

- $EQ_e$: Set of all equilibrium points on $e \in E$.

- $EQ$: Set of all equilibrium points on $G$.

### Example

```python
graph2 = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]
```

```python
ShowGraph = Graph(graph2)
ShowGraph.show()
```

```python
a = Local(graph2, 5, 1, 4)
print(a.EquiPoint(3, 2))
```

```python
graph3 = [
    [0, 2, 0, 6, 0, 0, 0, 0, 0, 1],
    [2, 0, 3, 8, 5, 0, 0, 0, 0, 0],
    [0, 3, 0, 0, 7, 4, 0, 0, 0, 0],
    [6, 8, 0, 0, 9, 0, 3, 0, 0, 0],
    [0, 5, 7, 9, 0, 0, 0, 2, 0, 0],
    [0, 0, 4, 0, 0, 0, 1, 6, 0, 0],
    [0, 0, 0, 3, 0, 1, 0, 0, 8, 0],
    [0, 0, 0, 0, 2, 6, 0, 0, 5, 3],
    [0, 0, 0, 0, 0, 0, 8, 5, 0, 7],
    [1, 0, 0, 0, 0, 0, 0, 3, 7, 0]
]
```

## Local center *[In progress...]*

### Description

For $e \in E, \: e = (v_i, v_j)$, the local center on $e$ is

$$
\arg \underset{x \in e}{\min} f(x) = \arg \min \{ \underset{x \in EQ_e}{\min} f(x), f(v_i), f(v_j) \}
$$

### Bounded value

```python
graph4 = [
    [0, 4, 3, 1, 2],
    [4, 0, 2, 4, 5],
    [3, 2, 0, 2, 3],
    [1, 4, 2, 0, 1],
    [2, 5, 3, 1, 0]
]
```

```python
ShowGraph = Graph(graph4)
ShowGraph.show()
```

Upper bound $f(v_i)$

```python
from packages.Point import LocalMinima
```

```python
a = LocalMinima(graph4, 5, 0, 3)

print(
    a.upperlm(1),
    a.upperlm(4),
    a.upperlm(2),
    a.upperlm(0)
)
```

```python
print(a.GeneralUpper())
```

Lower bound $f(v_j)$

```python
print(
    a.lowerlm(1),
    a.lowerlm(4),
    a.lowerlm(2),
    a.lowerlm(3)
)
```

```python
print(a.GeneralLower())
```
