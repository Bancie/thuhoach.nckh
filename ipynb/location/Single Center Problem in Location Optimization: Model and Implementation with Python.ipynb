# Introduction

## Bottleneck point

### Description

For $v \in V, \: x = (e, \alpha) \in G$, the bottleneck point $v$ on $e=(v_i,v_j)$ is

$$
\alpha l_e + d(v_i, v) = (1-\alpha ) l_e + d(v_j, v)
$$

- $BP_e$: Set of all bottleneck points on $e \in E$.

- $BP$: Set of all bottleneck points on $G$.

### Example

```python
graph = [
    [0, 1, 3],
    [1, 0, 2],
    [3, 2, 0]
]
```

```python
ShowGraph = Graph(graph)
ShowGraph.show()
```

```python
from packages.Point import Local
```

```python
a = Local(graph, 3, 0, 2)
print(a.BottleNeck(1))
```

## Equilibrium point

### Description

For $v_k, v_l \in V, \: v_k \neq v_l, \: x = (e, \alpha) \in G, \: e= (v_u, v_v)$, the equililbrium point is

$$
\omega _k (\alpha l_e+ d(v_k, v_u)) = \omega _l ((1-\alpha )l_e + d(v_l, v_v))
$$

If $x \in G$

$$
\omega _k d(v_k, x) = \omega _l d(v_l, x)
$$

- $EQ_e$: Set of all equilibrium points on $e \in E$.

- $EQ$: Set of all equilibrium points on $G$.

### Example

```python
graph2 = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]
```

```python
ShowGraph = Graph(graph2)
ShowGraph.show()
```

```python
a = Local(graph2, 5, 1, 4)
print(a.EquiPoint(3, 2))
```

## Local center

### Description

For $e \in E, \: e = (v_i, v_j)$, the local center on $e$ is

$$
\arg \underset{x \in e}{\min} f(x) = \arg \min \{ \underset{x \in EQ_e}{\min} f(x), f(v_i), f(v_j) \}
$$

### Bounded value

```python
graph4 = [
    [0, 4, 3, 1, 2],
    [4, 0, 2, 4, 5],
    [3, 2, 0, 2, 3],
    [1, 4, 2, 0, 1],
    [2, 5, 3, 1, 0]
]
ShowGraph = Graph(graph4)
ShowGraph.show()
```

Upper bound $f(v_i)$

```python
from packages.Point import LocalMinima
```

```python
a = LocalMinima(graph4, 5, 0, 1)

print(
    a.lowerlm(2),
    a.lowerlm(3),
    a.lowerlm(4),
    a.lowerlm(1)
)
```

```python
print(a.BoundedValue(x='upper'))
```

Lower bound $f(v_j)$

```python
print(
    a.upperlm(1),
    a.upperlm(4),
    a.upperlm(2),
    a.upperlm(3)
)
```

```python
print(a.BoundedValue(x='lower'))
```

### Interval value

$$
\underset{x \in EQ_e}{\min} f(x)
$$

The local minima on segment (A, D) (Not included lower and upper bound)

```python
eg = LocalMinima(graph4, 5, 0, 1)

min_val, min_x = eg.IntervalValue()
print(min_x)
print(min_val)
```

```python
eg.show_lines(1.5)
```

# Ordered Weber Region

```python
from packages.Point import FunctionPlotter
```

Example

```python
graph5 = [[0, 2, 1],
           [2, 0, 3],
           [1, 3, 0]
           ]

ord_weber = Graph(graph5)
ord_weber.show_node()
```

$$
x_1 = ((v_2, v_1), \alpha) =
\begin{cases}
f_1(x) &= 3 + 2x \\
f_2(x) &= 1 + (1-x)2
\end{cases}
$$

```python
BA = FunctionPlotter(0, 2, 100)
def f1(x): return 3 + 2 * x
BA.add_function(f1, "f1")
def f2(x): return 1 + (1 - x) * 2
BA.add_function(f2, "f2")
BA.find_intersection(0, 1)
BA.plot()
```

$$
x_2 = ((v_1, v_3), \alpha) =
\begin{cases}
f_3(x) &= 2 + \alpha \\
f_4(x) &= 3 + (1-\alpha)
\end{cases}
$$

```python
AC = FunctionPlotter(0, 1, 100)
def f3(x): return 2 + x
AC.add_function(f3, "f3")
def f4(x): return 3 + (1-x)
AC.add_function(f4, "f4")
AC.find_intersection(0, 1)
AC.plot()
```

$$
x_3 = ((v_3, v_2), \alpha) =
\begin{cases}
f_5 (x) &= 1 + 3 \alpha \\
f_6 (x) &= 2 + (1 - \alpha) 3
\end{cases}
$$

```python
CB = FunctionPlotter(0, 3, 100)
def f5(x): return 1 + 3*x
CB.add_function(f5, "f5")
def f6(x): return 2 + (1-x)*3
CB.add_function(f6, "f6")
CB.find_intersection(0, 1)
CB.plot()
```

```python
ord_weber.add_marked_point_on_edge('v2', 'v1', alpha=0, label='x1')
ord_weber.add_marked_point_on_edge('v1', 'v3', alpha=1, label='x2')
ord_weber.add_marked_point_on_edge('v3', 'v2', alpha=0.6667, label='x3')
ord_weber.show_node()
```
