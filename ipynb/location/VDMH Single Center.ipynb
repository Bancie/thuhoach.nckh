```python
import numpy as np
from scipy.sparse.csgraph import floyd_warshall
```

```python
g = [[0, 42, 53, 0],
    [42, 0, 39, 30],
    [53, 39, 0, 35],
    [0, 30, 35, 0]
    ]

graph = Graph(g)
graph.show_node()
```

```python
a = np.array(g, dtype=float)

# Replace 0s (excluding diagonal) with np.inf
a[a == 0] = np.inf
np.fill_diagonal(a, 0)

# Compute all-pairs shortest path matrix
g_shortest = floyd_warshall(a)

print(np.round(g_shortest, 2))
```

```python
graph = Graph(g_shortest)
graph.show_node()
```

$B_1 = d(v_1, v_2) =42$

$$

$$

```python
B1 = FunctionPlotter(0, 42, 100)
def f1(x): return x + 53
def f2(x): return 42 - x + 39
def f3(x): return x + 72
def f4(x): return 42 - x + 30
def f5(x): return x
def f6(x): return 42 - x
B1.add_function(f2, "f2 of d(v3,x)")
B1.add_function(f1, "f1 of d(v3,x)")
B1.add_function(f3, "f1 of d(v4,x)")
B1.add_function(f4, "f2 of d(v4,x)")
B1.add_function(f5, "d(v1,x)")
B1.add_function(f6, "d(v2,x)")
B1.find_intersection(0, 1)
B1.find_intersection(2, 3)
B1.find_intersection(1, 3)
B1.find_intersection(0, 4)
B1.plot()
```
