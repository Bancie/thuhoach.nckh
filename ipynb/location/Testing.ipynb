# Piecewise linear function

```python
from scipy.interpolate import PPoly
import numpy as np

# Example: y = x on [0,1], y = -x+2 on [1,2]
# Coefficients for each interval (from highest to lowest degree)
# Linear → degree 1 → shape: (2, n_intervals)
coefficients = [[1.0, -1.0],  # slopes (degree 1)
                [0.0, 2.0]]   # intercepts (degree 0)
breaks = [0.0, 1.0, 2.0]

pp = PPoly(coefficients, breaks)

# Evaluate at all breakpoints
values_at_breaks = pp(breaks)
print(values_at_breaks)  # Output: [0. 1. 0.]
```
