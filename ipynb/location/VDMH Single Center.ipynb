```python
import numpy as np
from scipy.sparse.csgraph import floyd_warshall
```

```python
g = [[0, 42, 53, 0],
    [42, 0, 39, 30],
    [53, 39, 0, 35],
    [0, 30, 35, 0]
    ]

graph = Graph(g)
graph.show_node()
```

```python
a = np.array(g, dtype=float)

# Replace 0s (excluding diagonal) with np.inf
a[a == 0] = np.inf
np.fill_diagonal(a, 0)

# Compute all-pairs shortest path matrix
g_shortest = floyd_warshall(a)

print(np.round(g_shortest, 2))
```

```python
graph = Graph(g_shortest)
graph.show_node()
```
